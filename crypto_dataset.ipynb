```python
clean_df = clean_df.sort_values(
    ['coin', 'date']
).reset_index(drop=True)
```

```python
clean_df[['coin', 'date', 'price']].head(10)
```

```python
clean_df['previous_price'] = (
    clean_df.groupby('coin')['price'].shift(1)
)
```

```python
clean_df['price_change_pct'] = (
    (clean_df['price'] - clean_df['previous_price'])
    / clean_df['previous_price']
) * 100
```

```python
clean_df['price_change'] = (
    clean_df['price'] - clean_df['previous_price']
)
```

```python
clean_df['previous_mkt_cap'] = (
    clean_df.groupby('coin')['mkt_cap'].shift(1)
)

clean_df['mkt_cap_change_pct'] = (
    (clean_df['mkt_cap'] - clean_df['previous_mkt_cap'])
    / clean_df['previous_mkt_cap']
) * 100
```

```python
def classify_performance(x):
    if pd.isna(x):
        return 'No Previous Data'
    elif x > 0:
        return 'Gainer'
    elif x < 0:
        return 'Loser'
    else:
        return 'No Change'

clean_df['performance'] = clean_df['price_change_pct'].apply(
    classify_performance
)
```
